# Jupyter 入門

コードを書く前に、Jupyter とは何か、なぜ Jupyter を使うのか、なぜ Python なのかを簡単に説明します。

Jupyter は3つの部分でできています。カーネル、Web サーバー、ノートブックアプリです。コードはすべてカーネルに送られ、カーネル上で実行されます。Web サーバーは、カーネルとノートブックの間でコードと実行結果をやり取りします。ノートブックアプリは、サーバー上で実行したコードの結果を表示するためのものです。コードの共有がしやすく、環境を統一できるなど、ほかにも多くの機能があります。詳しくは、[Project Jupyter の紹介ページ](https://jupyter.org)、[nature の記事](https://www.nature.com/news/interactive-notebooks-sharing-the-code-1.16261)、[O'Reilly の記事](https://www.oreilly.com/ideas/what-is-jupyter)、[Jupyter のドキュメント](https://jupyter-notebook-beginner-guide.readthedocs.io/en/latest/what_is_jupyter.html)をご覧ください。なお、Jupyter ノートブックで実行できるのは Python のコード**だけではありません**。

Python は数十年前からある言語ですが、[近年](https://bit.ly/2u1z6gh)人気が高まっています。そのため、まず Python で開発されるプロジェクトも増えています。Python には強いオープンソース文化もあります。
