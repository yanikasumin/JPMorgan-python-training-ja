# Jupyter 入門

コードに入る前に、Jupyter とは何か、なぜ Jupyter を使うのか、そしてなぜ Python なのかを説明します。

Jupyter は3つの部分から構成されています。カーネル、Web サーバー、そしてノートブックアプリです。実行されるコードはすべてカーネルに送られ、カーネル上で実行されます。Web サーバーは、カーネルとノートブックの間でコードと結果の送受信を処理します。そして、私たちにとって最も重要なノートブックアプリは、サーバー上で実行されたコードの結果を表示し、コードの共有を容易にし、統一された環境を提供します。このほかにも、さまざまな機能を備えています。興味のある方は、Jupyter とは何か、そして Jupyter がどのような影響を与えているかについて、[Project Jupyter のこちら](https://jupyter.org)、[nature のこちら](https://www.nature.com/news/interactive-notebooks-sharing-the-code-1.16261)、[O'Reilly のこちら](https://www.oreilly.com/ideas/what-is-jupyter)、[Jupyter のドキュメントのこちら](https://jupyter-notebook-beginner-guide.readthedocs.io/en/latest/what_is_jupyter.html)でさらに詳しく読むことができます。なお、Jupyter ノートブックが実行するのは Python のコード**だけではありません**。

Jupyter とは異なり、Python は数十年前から存在していますが、[近年](https://bit.ly/2u1z6gh)になって人気が高まってきました。その結果、いくつかのプロジェクトがまず Python で開発されるようになり、Python には強力なオープンソースの文化があります。
