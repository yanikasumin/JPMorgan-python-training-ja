# 科学計算のためのベストプラクティス

ベストプラクティスの要約です（[[1]](http://dx.doi.org/10.1371/journal.pbio.1001745) より）：

 1. プログラムは、コンピュータのためではなく、人のために書く。
 1. 作業はコンピュータに任せる。
 1. 変更は少しずつ行う。
 1. 繰り返しをしない（自分のコードも他人のコードも）。
 1. ミスが起きる前提で計画する。
 1. 最適化は、正しく動くようになってから行う。
 1. 文書に残すのは仕組みではなく、設計と目的。
 1. 共同作業をする。

# 1. コードの可読性

- 美しい方が、醜い方よりよい。

- 単純な方が、複雑な方よりよい。
- 複雑な方が、煩雑な方よりよい。
- 平らな方が、入れ子よりよい。
- 疎な方が、密な方よりよい。

## 変数名・関数名
 
科学者は、数式をそのままコードに書きがちです：
 
悪い例：
```python
p0 = 3.5
p = p0 * np.cos(0.4 * x - 13.2 * t)
```

良い例：
```python
# 定数:
base_pressure = 3.5  # Pa
wave_length = 15.7   # m
wave_number = 2 * np.pi / wave_length  # m-1
angular_frequency = 13.2  # Hz

# 計算:
pressure = base_pressure * np.cos(wave_number * x - angular_frequency * t)
```

*性能のために可読性を犠牲にする場合は、コードの内容を説明するコメントを必ず書いてください。*

## バージョン管理とは？

 - ファイルやフォルダの履歴を残す仕組みです。
 - 新しいバージョンを保存するときは、*コミット* を作ります。
 - バージョン管理しているフォルダを *リポジトリ* と呼びます。

## なぜバージョン管理を使うのか？

* バックアップ: 
    * スクリプト、ソースコード、ドキュメントなどの変更を管理できます
    * git リポジトリのコピーを外部のプラットフォーム（例： github）に置けます
* 整理:  
    * ファイルの古いバージョンを取り出せます。
    * 変更履歴を表示できます。
* 共同作業:
   * チームでのプログラミングや執筆に便利です。
   * 各自がリポジトリのファイルの *コピー* で作業し、変更を公式のリポジトリにアップロードします。
   * 別のメンバーの変更と競合しなければ、自動でマージされます。
   * 競合した変更は検出されるので、手作業で解決します。

## 仕組み

 - 単純なリポジトリでは、コミットが一列に並びます。
 - 変更が分岐して、後でマージされることもあります。

# 5. ツールを使ってテストを自動化する

プログラミングをすれば、ミスは必ずします。誰でもそうです。大切なのは、ミスを早く、効率よく見つけることです（リリースや公開の後に見つかるのは避けたいところです）。

利用できる最良のツール： ユニットテスト

## ユニットテストの書き方

 1. プログラムの中から、入力に対して決まった振る舞いをする *ユニット* を見つけます。ユニットとは、たとえば次のものです。
   - 関数
   - モジュール
   - プログラム全体
 1. ユニットに入力を与え、出力や振る舞いが期待どおりかを確認するテスト関数を書きます。
 1. テストは多ければ多いほどよいです。できれば複数のレベル（関数／モジュール／プログラム）で書いてください。

## Python でユニットテストを書く方法

[py.test](http://docs.pytest.org/en/latest/) や [nose](https://nose.readthedocs.io/en/latest/) のようなテストフレームワークを使います。ほかにもフレームワークはあります。

```bash
$ pip install pytest
```

`test_<ユニット名またはモジュール名>.py` というファイルを作ります。置き場所は `tests` というフォルダがおすすめです。

テスト対象のコードをインポートします。

`def test_<テスト名>():` という関数を追加します。その中で、入力に対して正しい振る舞いをするかを確認します。

`````python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

from mandlebrot import mandelbrot

x = np.linspace(-2.25, 0.75, 500)
y = np.linspace(-1.25, 1.25, 500)
output = mandelbrot(x, y, 200, False)

plt.imshow(output)
`````

`````python
import numpy as np
from mandlebrot import mandelbrot

def test_mandelbrot_small():
    x = np.linspace(-2.25, 0.75, 10)
    y = np.linspace(-1.25, 1.25, 10)
    output = mandelbrot(x, y, 100, False)
    assert output.shape == (10, 10)

def test_mandelbrot_zero_outside():
    # マンデルブロ集合は「アクティブ」な領域の外では 0 になる
    x = np.linspace(1.5, 2.0, 10)
    y = np.linspace(1.5, 2.0, 10)
    output = mandelbrot(x, y, 100, False)
    assert np.all(output == 0.0)

def test_mandelbrot_incorrect_test():
    x = np.linspace(-1.5, -2.0, 10)
    y = np.linspace(-1.25, 1.25, 10)
    output = mandelbrot(x, y, 100, False)
    assert np.all(output == 0.0)
  
`````

`````python
test_mandelbrot_small()
`````

`````python
test_mandelbrot_zero_outside()
`````

`````python
test_mandelbrot_incorrect_test()
`````

プロジェクトのフォルダで `py.test` または `nosetests` を実行します。ツールが、プロジェクト内のテスト（例： `test_*.py` ファイル内の `test_*()` 関数）をすべて探して実行します。

## 便利なテスト用ユーティリティ

`assert` と `==`、`>` などの比較演算子で、期待する振る舞いの大半は表現できます。ただ、テストをぐっと簡単に書けるユーティリティもあります。いくつか紹介します：

### NumPy のテスト

`````python
small_noise = 1e-6 * np.random.rand(10, 10)
np.testing.assert_almost_equal(small_noise, 0.0, decimal=5)  # 問題なし
np.testing.assert_almost_equal(small_noise, 0.0, decimal=7)  # おそらく失敗する
`````

## ローカルモジュール

 - これまでは、スクリプトやノートブックのセルにコードを書いてきました。
 - Python ファイルはどれもモジュールとして扱えます。つまり、他の Python ファイルやノートブックからインポートできます。

`````python
%%writefile mymodule.py

def my_function():
    return "Surprise!"
`````

この関数は、別のファイルやノートブックでインポートできます：

`````python
from mymodule import my_function

my_function()
`````
