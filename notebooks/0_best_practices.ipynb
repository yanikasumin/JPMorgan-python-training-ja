- 美しいものは、醜いものより良い。

- 単純なものは、複雑なものより良い。
- 複雑なものは、煩雑なものより良い。
- 平らなものは、入れ子になったものより良い。
- 疎なものは、密なものより良い。

- 今行うことは、決して行わないことより良い。

## 変数名・関数名
 
科学者はしばしば、自分たちの方程式をそのままコードに翻訳してしまいます：
 
悪い例：
```python
p0 = 3.5
p = p0 * np.cos(0.4 * x - 13.2 * t)
```

より良い例：
```python
# 定数:
base_pressure = 3.5  # Pa
wave_length = 15.7   # m
wave_number = 2 * np.pi / wave_length  # m-1
angular_frequency = 13.2  # Hz

# 計算:
pressure = base_pressure * np.cos(wave_number * x - angular_frequency * t)
```

*性能上の理由で可読性を犠牲にしなければならない場合は、コードの内容を説明するコメントを必ず追加してください。*

## バージョン管理とは？

 - ファイル／フォルダの履歴を保存するためのシステムです。
 - 通常、ファイルの新しいバージョンを保存したいときは、*コミット* を行います。
 - バージョン管理下にあるフォルダは、通常 *リポジトリ* と呼ばれます。

## なぜバージョン管理を使うのか？

* バックアップ: 
    * スクリプト、ソースコード、ドキュメントなどのファイルの変更を管理できます
    * git リポジトリのコピーを外部のプラットフォーム（例： github）に保存できます
* 整理:  
    * ファイルの古いバージョンを取り出せます。
    * 変更の履歴を表示できます。
* 共同作業:
   * チームでのプログラミングや執筆に役立ちます。
   * プログラマーはリポジトリのファイルの *コピー* を使って作業し、変更を公式のリポジトリにアップロードします。
   * 異なるチームメンバーによる競合しない変更は、自動的にマージされます。
   * 競合する変更は検出され、手作業で解決しなければなりません。

## どのように動作するのか？

 - 単純なリポジトリは、通常、コミットが直線的に連なったものになります。
 - ただし、変更が分岐し、再びマージされることもあります。

# 5. ツールを使ってテストを自動化する

プログラミングをしていれば、ミスは必ずします！ 誰でもそうです。大切なのは、ミスを早期に、効率的に発見することです（リリース／公開の後では、通常、望ましくありません）。

利用できる最良のツール： ユニットテスト

## ユニットテストの書き方

 1. プログラムの中で、特定の入力に対して明確に定義された振る舞いをするはずの *ユニット* を特定します。ユニットには次のようなものがあります。
   - 関数
   - モジュール
   - プログラム全体
 1. 特定の入力をユニットに与え、出力や振る舞いが期待どおりであることを確認するテスト関数を書きます
 1. 多ければ多いほど良いです！ できれば複数のレベル（関数／モジュール／プログラム）で書いてください。

## Python でユニットテストを書く方法

[py.test](http://docs.pytest.org/en/latest/) や [nose](https://nose.readthedocs.io/en/latest/) のようなテストフレームワークを使ってください。他にもいくつかのフレームワークが存在します。

```bash
$ pip install pytest
```

`tests` という名前のフォルダの中に、`test_<ユニット名またはモジュール名>.py` というファイルを作成してください。

テスト対象のコードをインポートします。

`def test_<テスト名>():` という関数を追加し、特定の入力に対して正しい振る舞いをするかどうかを確認させます

`````python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

from mandlebrot import mandelbrot

x = np.linspace(-2.25, 0.75, 500)
y = np.linspace(-1.25, 1.25, 500)
output = mandelbrot(x, y, 200, False)

plt.imshow(output)
`````

`````python
import numpy as np
from mandlebrot import mandelbrot

def test_mandelbrot_small():
    x = np.linspace(-2.25, 0.75, 10)
    y = np.linspace(-1.25, 1.25, 10)
    output = mandelbrot(x, y, 100, False)
    assert output.shape == (10, 10)

def test_mandelbrot_zero_outside():
    # マンデルブロ集合は「アクティブ」な領域の外側では 0 になるはずです
    x = np.linspace(1.5, 2.0, 10)
    y = np.linspace(1.5, 2.0, 10)
    output = mandelbrot(x, y, 100, False)
    assert np.all(output == 0.0)

def test_mandelbrot_incorrect_test():
    x = np.linspace(-1.5, -2.0, 10)
    y = np.linspace(-1.25, 1.25, 10)
    output = mandelbrot(x, y, 100, False)
    assert np.all(output == 0.0)
  
`````

`````python
test_mandelbrot_small()
`````

`````python
test_mandelbrot_zero_outside()
`````

`````python
test_mandelbrot_incorrect_test()
`````

## 期待される振る舞いを表現するためのユーティリティ

`assert` や `==`、`>` などの比較演算子で、期待される振る舞いの大半は表現できるはずですが、テストを大幅に簡略化できるユーティリティがいくつかあります。そのうちのいくつかを紹介します：

### NumPy のテスト

`````python
small_noise = 1e-6 * np.random.rand(10, 10)
np.testing.assert_almost_equal(small_noise, 0.0, decimal=5)  # 問題ありません
np.testing.assert_almost_equal(small_noise, 0.0, decimal=7)  # おそらく失敗します
`````

## ローカルモジュール

 - これまでは、スクリプトやノートブックのセルにだけコードを書いてきました。
 - すべての Python ファイルはモジュールとして扱うこともできます。つまり、他の Python ファイル／ノートブックからインポートできます。
 - ただし、スクリプトと比べると、モジュールは通常、どこか別の場所でインポートされることを意図した定義（関数／クラス／定数）のみを含みます

`my_function()` という関数を持つファイル `mymodule.py` があるとします：

`````python
%%writefile mymodule.py

def my_function():
    return "Surprise!"
`````

その関数は、別のファイルやノートブックでインポートできます：

`````python
from mymodule import my_function

my_function()
`````
